```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
import warnings 
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("cleaned_data.csv")
df.head()
```

```python
df.columns
```

## Creating the dataframe only for the required columns for the model prediction:-

```python
df1 = df[["average_cost_for_two","price_range","votes","has_table_booking","has_online_delivery","is_delivering_now","country_code","city","cuisines"]]
df1["aggregate_rating"] = df["aggregate_rating"]
df1.head(10)
```

```python
df1 = df1.drop(["city","cuisines"],axis = 1)
df1.head(5)
```

## **Insights:-**
- `we kept the only columns that we need for making model.`

## Defining the freatures and target:-


```python
x = df1.drop("aggregate_rating",axis = 1)
y = df["aggregate_rating"]
```

```python
print(x.shape)
```

## **Insights:-**
- `here we seperated the features and the target`.

## Spliting the fratures and the target data into train Test split :-

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

## **Encoding:-** (`labelencoding`)

```python
# Convert Yes/No into 1/0

binary_cols = [
    'has_table_booking',
    'has_online_delivery',
    'is_delivering_now'
]

for col in binary_cols:
    x_train[col] = x_train[col].map({'Yes': 1, 'No': 0})
    x_test[col] = x_test[col].map({'Yes': 1, 'No': 0})
```

```python
x.head(5)
```

## **Insights :-**
- `here we can see the all our data is in the form of numbers.`

```python
x.shape
```

```python
x
```

## Importing Algorithm :- (`Gradientbossting & RandomForest`)

```python
from sklearn.ensemble import GradientBoostingRegressor,RandomForestRegressor
```

## Model making with hypertuning parameters:- 

```python
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

## **Insights:-** 
- here we set the parameters to make our model best to predict the values of average rating.
- `n_estimators=100`
- `random_state=42`
- `n_jobs=-1`

## Training model on tranining data:- 

```python
model.fit(x_train, y_train)
```

### Model trained successfully

## **Making Predictions :-**

```python
model_pred = model.predict(x_test)
pd.DataFrame({
    "Actual prediction " : y_test,
    "model Prediction"  : model_pred.round(2)
})
```

## Evaluating the model:-

```python
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
```

```python
print("MAE:", mean_absolute_error(y_test, model_pred))
print("MSE:", mean_squared_error(y_test, model_pred))
print("RMSE:", mean_squared_error(y_test, model_pred)**0.5)
print("R2 Score:", r2_score(y_test, model_pred))
```

## Model making with hypertuning parameters:- 

```python
model2 = GradientBoostingRegressor(
    n_estimators = 100,
    random_state = 42,
    max_depth = 5,
)
model2.fit(x_train,y_train)
```

## **Insights:-** 
- here we set the parameters to make our model best to predict the values of average rating.
- `n_estimators=100`
- `random_state=42`
- `max_depth = 5`

## **Making predictions:-**

```python
model2_pred = model2.predict(x_test)
pd.DataFrame({
    "Actual Prediction " : y_test,
    "model Prediction " : model2_pred.round(2)
})
```

## Evaluating the model:-

```python
print("MAE:",mean_absolute_error(y_test,model2_pred))
print("MSE:",mean_squared_error(y_test,model2_pred))
print("RMSE",mean_squared_error(y_test,model2_pred)**0.5)
print("R2_score:",r2_score(y_test,model2_pred))
```

## **Insights:-** 
- The Gradient Boosting Regressor achieved an R² score of 95.18%, with an MAE of 0.2164 and RMSE of 0.3313. 
- These results indicate strong predictive performance with relatively low prediction errors. 
- Compared with Random Forest, Gradient Boosting delivered slightly better results and was the best-performing model among the two algorithms evaluated so far.
